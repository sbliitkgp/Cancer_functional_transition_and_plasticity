```r
library("R.matlab")
library("stringr")
sample_name <- "BRONCHO_58_all_genes"

parti_output_path <- paste("../../functional_heterogeneity_pipeline/Parti_result", sample_name, sep="/")
files <- list.files(parti_output_path)        
parti_output_file <- files[str_detect(files, ".mat$")]

parti_output <- readMat(paste(parti_output_path, parti_output_file, sep="/"))

pc_data <- as.data.frame(parti_output$pc)
pc_data <- pc_data[,1:2]
colnames(pc_data) <- c("pc1", "pc2")

celltype_folder <- list.files(path = paste("../../functional_heterogeneity_pipeline/preprocessing", sample_name,sep="/"))
celltype_data_file <- celltype_folder[str_detect(celltype_folder, ".tsv$") & str_detect(celltype_folder, "modified")]
cell_type <- read.table(paste(paste("../../functional_heterogeneity_pipeline/preprocessing", sample_name,sep="/"),celltype_data_file,sep="/"),sep='\t',na.strings =c("","NA"), header = T)
rownames(pc_data) <- cell_type$Index
pc_data <-  cbind(pc_data, cell_type$Cell_type.refined)
colnames(pc_data)[3] <- "cell_type"

arc <- as.data.frame(parti_output$arc)
arc <- arc[,1:2]
colnames(arc) <- c("pc1", "pc2")

sample <- word(sample_name, 1, 2, sep = "_")
mellon_folder <- paste("../../functional_heterogeneity_pipeline/mellon_analysis/", sample, sep="/")
mellon_files <- list.files(mellon_folder)
cell_file <- mellon_files[str_detect(mellon_files, ".csv$") & str_detect(mellon_files, "pareto_front")]
mellon_cells <- read.csv(paste(mellon_folder, cell_file, sep="/"), row.names=1)

pc_data_final <- pc_data[rownames(pc_data) %in% rownames(mellon_cells),]

library("ggplot2")

export_path <- "../../functional_heterogeneity_pipeline/celltype_coloring/mellon_celltype_coloring"
dir.create(paste(export_path, sample, sep="/"), recursive = TRUE)

pdf(file = paste(export_path,sample,paste(sample,"2d.pdf",sep=" "),sep="/"),height=20,width=30)
plot_cell_type <- ggplot(pc_data_final, aes(x=pc1,y=pc2,color=cell_type),size=20)+
  geom_point(size = 6)+
  geom_point(data=arc,aes(x=pc1,y=pc2),color='black',size=15)+
  guides(color = guide_legend(override.aes = list(size = 20)))+
  ggtitle(sample)+
  labs(x="PC1",y="PC2", color="Cell Types")+
  theme_classic()+
  theme(
    axis.line = element_line(color = "black", linewidth = 2.5),
    axis.ticks = element_line(color = "black", linewidth = 2),
    axis.text = element_text(color = "black", size = 80),
    axis.ticks.length = unit(0.2, "cm")
  )
  
plot_cell_type + scale_color_manual(values=c("T/NK cells" = "#0000FF", "Myeloid cells" = "#c90076", "Epithelial cells" = "#008000"))+  
  theme(text = element_text(size = 95)) + annotate("text", x = arc[,1]+3, y = arc[,2]+3, label = rownames(arc),size=32, fontface="bold")
dev.off()
```
