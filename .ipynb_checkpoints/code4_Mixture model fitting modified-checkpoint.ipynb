```python
## shifting from python kernel to R kernel
library(glue)
library(multimode)
library(mixtools)
library(MASS)

dir.create("../../pipeline_testing/distance_distribution/", recursive=T)

i=1 # Please mention the archetype number, if it is "Archetype 1' then put 1, if it 'Archetype 2' then put 2, like that 
folder_name <- "SMC01-T" # Please put the number of archetypes present here in this respective sample
data <- read.csv(glue("../../pipeline_testing/distance_and_ordering/{folder_name}/cell_distances.csv"), row.names=1)
df <- as.numeric(data[i,])

print(modetest(df))

nmodes <- c(1,2,3,4,5,6) #For this study we have considered upto 6 modes. Unnecessary extra number of modes most of time overfits the model.

for (j in 1:length(nmodes)){
    if (nmodes[j]>1){
        # #finding modes (Maximum frequency) and antimodes (least frequent value between two modes)
        result <- locmodes(df,mod0=nmodes[j],display=TRUE,posLegend="topright")
        print(result$locations)
        
        ## Fitting the data with more than one normal distributions
        fit <- normalmixEM(df,k=nmodes[j])
        mu <- sort(fit$mu)
        sigma <- sort(fit$sigma)
        lambda <- sort(fit$lambda)
        fit_result <- as.data.frame(cbind(mu,sigma,lambda))
        write.csv(fit_result, paste(glue("../../pipeline_testing/distance_distribution/{folder_name}"), glue("Archetype {i}/norm_dist_param_nmodes_{j}.csv") ,sep="/"))
    }else if (nmodes[j]==1){
        fit <- fitdistr(df, "normal")
        para <- fit$estimate
        para_mu<-para[1]
        para_sigma <- para[2]
        fit_result_unimod <- as.data.frame(cbind(para_mu,para_sigma))
        write.csv(fit_result_unimod, paste(glue("../../pipeline_testing/distance_distribution/{folder_name}"), glue("Archetype {i}/norm_dist_param_nmodes_{j}.csv") ,sep="/"))
    }
}
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
import seaborn as sns
from lmfit import Model
import pathlib

folder_name = "SMC01-T" # Plese mention your sample names here
data = pd.read_csv(f'../../pipeline_testing/distance_and_ordering/{folder_name}/cell_distances.csv',index_col=[0])

i=1 # Please mention the archetype number, if it is "Archetype 1' then put 1, if it 'Archetype 2' then put 2, like that
nmodes = [1,2,3,4,5,6]
new_dir_name = f"Archetype {i}"
new_dir = pathlib.Path(f"../../pipeline_testing/distance_distribution/{folder_name}", new_dir_name)
new_dir.mkdir(parents=True, exist_ok=True)
new_file = new_dir/f'Archetype {i} fitted distance distribution.txt'

all_fit_result = []
for mode in nmodes:
    fit_result = pd.read_csv(f"../../pipeline_testing/distance_distribution/{folder_name}/{new_dir_name}/norm_dist_param_nmodes_{mode}.csv",index_col=0)
    all_fit_result.append(fit_result)

df= data.iloc[i-1,:]
y,x,_ = plt.hist(df,bins=50,histtype="bar",alpha=0.3,density=True,edgecolor="black")
x=(x[1:]+x[:-1])/2

def gauss(x, mu, sigma, A):
    return A*np.exp(-(x-mu)**2/(2*sigma**2))

def bimodal(x, mu1, sigma1, A1, mu2, sigma2, A2):
    return gauss(x,mu1,sigma1,A1)+gauss(x,mu2,sigma2,A2)

def trimodal(x, mu1, sigma1, A1, mu2, sigma2, A2,mu3,sigma3,A3):
     return gauss(x,mu1,sigma1,A1)+gauss(x,mu2,sigma2,A2)+gauss(x,mu3,sigma3,A3)
    
def tetramodal(x, mu1, sigma1, A1, mu2, sigma2, A2, mu3, sigma3, A3, mu4, sigma4, A4):
     return gauss(x,mu1,sigma1,A1)+gauss(x,mu2,sigma2,A2)+gauss(x,mu3,sigma3,A3)+gauss(x,mu4,sigma4,A4)

def pentamodal(x, mu1, sigma1, A1, mu2, sigma2, A2, mu3, sigma3, A3, mu4, sigma4, A4, mu5, sigma5, A5):
     return gauss(x,mu1,sigma1,A1)+gauss(x,mu2,sigma2,A2)+gauss(x,mu3,sigma3,A3)+gauss(x,mu4,sigma4,A4)+gauss(x,mu5,sigma5,A5)

def hexamodal(x, mu1, sigma1, A1, mu2, sigma2, A2, mu3, sigma3, A3, mu4, sigma4, A4, mu5, sigma5, A5, mu6, sigma6, A6):
     return gauss(x,mu1,sigma1,A1)+gauss(x,mu2,sigma2,A2)+gauss(x,mu3,sigma3,A3)+gauss(x,mu4,sigma4,A4)+gauss(x,mu5,sigma5,A5)+gauss(x,mu6,sigma6,A6) 

def model_fitting(df): 
    if (df.shape[0]==1):
        gmodel = Model(gauss)
        result_model = gmodel.fit(y, x=x, A=0.8, mu=df.iloc[0,0], sigma=df.iloc[0,1])
        cut_off_val = result_model.params['mu'].value - abs(result_model.params['sigma'].value)
        
    elif (df.shape[0]==2):
        gmodel = Model(bimodal) 
        result_model = gmodel.fit(y, x=x, A1=df.iloc[0,2], mu1=df.iloc[0,0], sigma1=df.iloc[0,1], 
                                    A2=df.iloc[1,2], mu2=df.iloc[1,0], sigma2=df.iloc[1,1])  
        mus = [result_model.params['mu1'].value, result_model.params['mu2'].value]
        lowest_mu = min(mus)
        best_idx = mus.index(lowest_mu) + 1
        sigma = result_model.params[f'sigma{best_idx}'].value
        cut_off_val = lowest_mu + abs(sigma)
        
    elif (df.shape[0]==3):
        gmodel = Model(trimodal) 
        result_model = gmodel.fit(y, x=x, A1=df.iloc[0,2], mu1=df.iloc[0,0], sigma1=df.iloc[0,1], 
                                    A2=df.iloc[1,2], mu2=df.iloc[1,0], sigma2=df.iloc[1,1], 
                                    A3=df.iloc[2,2], mu3=df.iloc[2,0], sigma3=df.iloc[2,1])
        mus = [result_model.params['mu1'].value, result_model.params['mu2'].value, result_model.params['mu3'].value]
        lowest_mu = min(mus)
        best_idx = mus.index(lowest_mu) + 1
        sigma = result_model.params[f'sigma{best_idx}'].value
        cut_off_val = lowest_mu + abs(sigma)
        
    elif (df.shape[0]==4):
        gmodel=Model(tetramodal)
        result_model = gmodel.fit(y, x=x, A1=df.iloc[0,2], mu1=df.iloc[0,0], sigma1=df.iloc[0,1], 
                                    A2=df.iloc[1,2], mu2=df.iloc[1,0], sigma2=df.iloc[1,1], 
                                    A3=df.iloc[2,2], mu3=df.iloc[2,0], sigma3=df.iloc[2,1], 
                                    A4=df.iloc[3,2], mu4=df.iloc[3,0], sigma4=df.iloc[3,1])
        mus = [result_model.params['mu1'].value, result_model.params['mu2'].value, result_model.params['mu3'].value,result_model.params['mu4'].value]
        lowest_mu = min(mus)
        best_idx = mus.index(lowest_mu) + 1
        sigma = result_model.params[f'sigma{best_idx}'].value
        cut_off_val = lowest_mu + abs(sigma)
    
    elif (df.shape[0]==5):
        gmodel=Model(pentamodal)
        result_model = gmodel.fit(y, x=x, A1=df.iloc[0,2], mu1=df.iloc[0,0], sigma1=df.iloc[0,1], 
                                    A2=df.iloc[1,2], mu2=df.iloc[1,0], sigma2=df.iloc[1,1], 
                                    A3=df.iloc[2,2], mu3=df.iloc[2,0], sigma3=df.iloc[2,1], 
                                    A4=df.iloc[3,2], mu4=df.iloc[3,0], sigma4=df.iloc[3,1],
                                    A5=df.iloc[4,2], mu5=df.iloc[4,0], sigma5=df.iloc[4,1])
        mus = [result_model.params['mu1'].value, result_model.params['mu2'].value, result_model.params['mu3'].value,result_model.params['mu4'].value,
               result_model.params['mu5'].value]
        lowest_mu = min(mus)
        best_idx = mus.index(lowest_mu) + 1
        sigma = result_model.params[f'sigma{best_idx}'].value
        cut_off_val = lowest_mu + abs(sigma)

    elif (df.shape[0]==6):
        gmodel=Model(hexamodal)
        result_model = gmodel.fit(y, x=x, A1=df.iloc[0,2], mu1=df.iloc[0,0], sigma1=df.iloc[0,1], 
                                    A2=df.iloc[1,2], mu2=df.iloc[1,0], sigma2=df.iloc[1,1], 
                                    A3=df.iloc[2,2], mu3=df.iloc[2,0], sigma3=df.iloc[2,1], 
                                    A4=df.iloc[3,2], mu4=df.iloc[3,0], sigma4=df.iloc[3,1],
                                    A5=df.iloc[4,2], mu5=df.iloc[4,0], sigma5=df.iloc[4,1],
                                    A6=df.iloc[5,2], mu6=df.iloc[5,0], sigma6=df.iloc[5,1])
        mus = [result_model.params['mu1'].value, result_model.params['mu2'].value, result_model.params['mu3'].value,result_model.params['mu4'].value,
               result_model.params['mu5'].value, result_model.params['mu6'].value]
        lowest_mu = min(mus)
        best_idx = mus.index(lowest_mu) + 1
        sigma = result_model.params[f'sigma{best_idx}'].value
        cut_off_val = lowest_mu + abs(sigma)
        
    return cut_off_val, result_model


final_output =[model_fitting(d) for d in all_fit_result]

min_x = min(x)
max_x = max(x)

# 1. Filter out models where the cut-off is outside the data range
valid_outputs =[]
for original_index, (cutoff, model) in enumerate(final_output):
    if min_x <= cutoff <= max_x:
        valid_outputs.append((original_index, (cutoff, model)))
        
if len(valid_outputs) == 0:
    print("Warning: ALL models have cut-offs outside the data range! Ignoring the range rule.")
    valid_outputs = list(enumerate(final_output))

# 2. Find the best model
best_index, best_tuple = min(valid_outputs, key=lambda item: item[1][1].aic)

best_cut_off_value = best_tuple[0]
best_model = best_tuple[1]
print(f"The best overall model came from dataframe index: {best_index}")
print(f"The best model has a AIC of: {best_model.aic}")

plt.style.use('seaborn-v0_8-white')
plt.plot(x, y,'o')
plt.xlabel('Distance from Archetype',fontsize=10)
plt.ylabel('Density',fontsize=10)
plt.title(f'Fitted model for distance distribution from Archetype {i}',fontsize=14)
plt.plot(x, best_model.best_fit, '-', label='best fit')
plt.axvline(x = best_cut_off_value, color = 'red', linewidth = 2)
plt.legend(prop = { "size": 7}, loc ="upper right")
plt.savefig('{}/fitted model of distance distribution.pdf'.format(new_dir),dpi=300)

final_result = best_model.fit_report()

with open(new_file, 'w') as f:
    f.write(str(final_result))   
```
