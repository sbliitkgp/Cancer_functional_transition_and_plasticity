```r
path <- getwd()
setwd(path)
library(stringr)
sample <- "EBUS_49"
archetype_path <- paste("/data/subhasis_backup/subhasis/functional_heterogeneity_pipeline/specialist_generalist_cell_population_info(according to PARTI)/without_cluster",sample,sep="/")
archetype_files <- list.files(path = archetype_path)

#marged cells
specialist_generalist <- list()
for(i in 1:length(archetype_files)){
    data <- read.table(paste(archetype_path,archetype_files[i],sep="/"),header=FALSE,sep="\t")
    specialist_generalist[[i]] <- data
}
gene_exp_path <- paste("/data/subhasis_backup/subhasis/functional_heterogeneity_pipeline/preprocessing",paste(sample,"_all_genes",sep=""),sep="/")
gene_exp_file <- list.files(path= gene_exp_path)
TPM_gene_exp_file <- gene_exp_file[str_detect(gene_exp_file, ".csv$") & str_detect(gene_exp_file, "TPM_mod")]
tpm_data <- read.csv(paste(gene_exp_path, TPM_gene_exp_file,sep="/"),row.names=1)

specialist_generalist_exp <- list()
for (j in 1:length(specialist_generalist)){
    specialist_generalist_exp[[j]] <- tpm_data[,match(specialist_generalist[[j]]$V1,colnames(tpm_data))]
}

specialist_generalist_exp_total <- cbind(specialist_generalist_exp[[1]],specialist_generalist_exp[[2]],specialist_generalist_exp[[3]],
                                         specialist_generalist_exp[[4]],specialist_generalist_exp[[5]],specialist_generalist_exp[[6]])

specialist_generalist_data <- specialist_generalist_exp_total[rowSums(specialist_generalist_exp_total)>=1,]

#calculate varience across rows
specialist_generalist_data$varience <- apply(specialist_generalist_data[,-1], 1, var)
min(specialist_generalist_data$varience)
max(specialist_generalist_data$varience)
heatmap_data <- specialist_generalist_data[specialist_generalist_data$varience>1,]
heatmap_data <- heatmap_data[,1:ncol(heatmap_data)-1]
print(paste("up_deg heatmap input data dimension",dim(heatmap_data)))
matrix_heatmap <- as.matrix(heatmap_data)

library(gplots)
#png(filename="heatmap_specialists.png",width=1800,height=1800,units="px")
#my_palette <- colorRampPalette(c("red", "blue", "green"))(n = 100)
#breaks <- c(-50, -20, 20, 50)

pdf(file=paste(sample,"heatmap_specialist_&_generalist.pdf",sep=" "), width=20, height=20)
par(mar = c(1, 4, 2, 4))  # Adjust the bottom margin value as needed

heatmap.2(matrix_heatmap,
          trace = "none",
          Colv = FALSE,
          Rowv = TRUE,
          dendrogram = 'none',
          cexRow = 1,
          cexCol = 0.7,
          keysize = 0.8,
          col = bluered(256),
          margins = c(25, 12),
          ColSideColors = c(
            rep("orange", ncol(specialist_generalist_exp[[1]])),  
            rep("grey", ncol(specialist_generalist_exp[[2]])),
            rep("green", ncol(specialist_generalist_exp[[3]])),
            rep("yellow", ncol(specialist_generalist_exp[[4]])),
            rep("blue", ncol(specialist_generalist_exp[[5]])),
            # rep("purple", ncol(specialist_generalist_exp[[6]])),
            rep("brown", ncol(specialist_generalist_exp[[6]]))  
          ),
          scale = 'row',
          breaks = seq(-3, 3, length.out = 257))

legend("topright", 
       legend = c("Archetype 1", "Archetype 2", "Archetype 3", "Archetype 4", "Archetype 5", "Generalists"),
       fill = c('orange', 'grey', 'green', 'yellow', 'blue', 'brown'), 
       border = FALSE, 
       bty = "n", 
       y.intersp = 0.7, 
       cex = 2)

dev.off()
```
