```python
from pathlib import Path
import re

# 1. Define the folder path
folder_path = Path("../../functional_heterogeneity_pipeline/mellon_analysis/LUNG_T34/")

# 2. Iterate through all .csv files
for file_path in folder_path.glob("*.csv"):
    
    # 3. String Manipulation
    # - Convert to lowercase (optional, but recommended for consistency)
    # - Remove the word 'hallmark'
    # - Replace spaces with underscores
    new_name = file_path.name.lower().replace("hallmark", "")
    new_name = new_name.replace(" ", "_")
    
    # 4. Clean up underscores
    # If the original was "test hallmark data.csv", it might now be "test__data.csv"
    # This regex replaces multiple underscores with a single one
    new_name = re.sub(r'_+', '_', new_name)
    
    # Remove leading or trailing underscores (e.g., "_apoptosis.csv" -> "apoptosis.csv")
    new_name = new_name.strip("_")
    
    # 5. Define the new full path and rename
    new_file_path = file_path.with_name(new_name)
    
    if file_path != new_file_path:
        file_path.rename(new_file_path)
        print(f"Renamed: {file_path.name} -> {new_name}")

print("Cleanup complete!")
```

```python
import os
import re

def clean_preserve_case(text):
    # 1. Remove 'hallmark' regardless of case (Hallmark, HALLMARK, etc.)
    # We use re.sub with flags=re.IGNORECASE to find it without changing the whole string to lowercase
    text = re.sub(r'hallmark', '', text, flags=re.IGNORECASE)
    
    # 2. Replace spaces and hyphens with underscores
    text = text.replace(" ", "_").replace("-", "_")
    
    # 3. Collapse multiple underscores (___) into one (_)
    text = re.sub(r'_+', '_', text)
    
    # 4. Remove underscores from the very start or end
    text = text.strip("_")
    
    return text

def rename_everything(root_path):
    # topdown=False is necessary so we don't lose the path to files 
    # when we rename the folder they are in.
    for root, dirs, files in os.walk(root_path, topdown=False):
        
        # --- RENAME FILES ---
        for name in files:
            if name.lower().endswith(".csv"):
                # Split extension so we don't rename '.csv' to '.CSV' or similar
                base, ext = os.path.splitext(name)
                new_base = clean_preserve_case(base)
                
                old_path = os.path.join(root, name)
                new_path = os.path.join(root, new_base + ext)
                
                if old_path != new_path:
                    os.rename(old_path, new_path)
                    print(f"File: {name} -> {new_base + ext}")

        # --- RENAME SUBFOLDERS ---
        for name in dirs:
            new_name = clean_preserve_case(name)
            
            old_path = os.path.join(root, name)
            new_path = os.path.join(root, new_name)
            
            if old_path != new_path:
                try:
                    os.rename(old_path, new_path)
                    print(f"Folder: {name} -> {new_name}")
                except Exception as e:
                    print(f"Error renaming folder {name}: {e}")

# --- Set your path here ---
target_directory = "/home/subhasis/functional_heterogeneity_pipeline/GSEA/mellon_gavish"
rename_everything(target_directory)
```
