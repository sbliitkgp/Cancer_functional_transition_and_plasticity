# Performing UMAP for dimensionality reduction of specialist-genetalist gene expression data

```python
import umap
specialist_generalist_gene_exp_transposed = specialist_generalist_gene_exp.T
df = specialist_generalist_gene_exp_transposed.iloc[:,0:len(specialist_generalist_gene_exp_transposed.columns)]
embedding = umap.UMAP(n_neighbors=5,min_dist=0,n_components=3,random_state=42).fit_transform(df.values) ## Please change the n_neighbors according to your need and also keep that n-neighbors where you get better silhoutte score
embedding.shape
```

```python
#Caclculate trustworthiness score of umap result
import sklearn.manifold as manifold
score = manifold.trustworthiness(df.values, embedding)
score
```

```python
import matplotlib.pyplot as plt

embedding_df = pd.DataFrame(embedding)
embedding_df['Archetype_no'] = specialist_generalist_cells["Archetype_no"]

color_map = {
         'Archetype 2': 'cyan',
       # 'Archetype 3': 'olivedrab',
       # 'Archetype 4': 'seagreen',
       # 'Archetype 5': 'purple'  
    }
colors = embedding_df['Archetype_no'].map(color_map)
plt.scatter(embedding_df.iloc[:, 0], embedding_df.iloc[:, 1],c=colors, s=10)
plt.title('UMAP clustering of specialist-generalist cell population', fontsize=20)
plt.xlabel('UMAP_1')
plt.ylabel('UMAP_2')
plt.show()
```

```python
new_df = specialist_generalist_gene_exp_transposed.reset_index()
new_df = new_df.rename(columns={new_df.columns[0]: 'cell barcode'})
new_df.head(n=30)
```

# Applying HDBSCAN on UMAP reduced data

```python
import hdbscan
import sklearn.metrics as metrics
cluster_size = int((len(specialist_generalist_gene_exp_transposed)*3)/100)
clusterer = hdbscan.HDBSCAN(min_cluster_size=cluster_size, allow_single_cluster=True)
clusterer.fit(embedding)

# get cluster labels and add it to the data
print(clusterer.labels_)
df_mod = pd.DataFrame(embedding)
df_mod["cell_barcode"]=new_df["cell barcode"]
df_mod["label"]=clusterer.labels_
# df_mod["Archetype_no"] = new_df["Archetype_no"]
df_mod["Archetype_no"] = embedding_df["Archetype_no"]

sh_score = metrics.silhouette_score(embedding, clusterer.labels_)
print("Silhoutte score: %0.4f" %sh_score)
```
